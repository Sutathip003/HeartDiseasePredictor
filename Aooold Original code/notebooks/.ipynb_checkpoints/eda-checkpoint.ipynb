#### Feature Description
- id: Unique id for each patient
- age: Age of the patient
- sex: Gender (1 = male, 0 = female)
- dataset: location of data collection
- cp: Chest pain type (asymptomatic,non-anginal,atypical angina,non-anginal)
- trestbps: Resting blood pressure → measured with blood pressure cuff
- chol: Serum cholesterol (mg/dl)
- fbs: fasting blood sugar (True, False)
- restecg: (resting electrocardiographic results) → ECG at rest
- thalach: Maximum heart rate achieved → measured during treadmill test 
- exang: exercise-induced angina (True/ False) → patient reports pain during exercise
- oldpeak: ST depression induced by exercise relative to rest → ECG change during exercise → shape of ECG curve
- slope: the slope of the peak exercise ST segment → doctor asks patient about chest pain
- ca: number of major vessels (0-3) colored by fluoroscopy → number of blocked vessels
- thal:[normal; fixed defect; reversible defect]  → blood flow (normal / defect)
- num: the predicted attribute → doctor labels patient (disease / no disease)

# STEP 2: STRUCTURAL ANALYSIS

```python
# Data basic info
print("Dataset Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
```

```python
#Identify numerical vs categorical (basic)
numerical = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical = df.select_dtypes(include=['object']).columns.tolist()

print("Initial Numerical:", numerical)
print("Initial Categorical:", categorical)
```

```python
# Remove ID and Target form list
id_cols = ['id']
target_col = 'num'
id_target_col = id_cols + [target_col]

numerical = [col for col in numerical if col not in id_target_col]
categorical = [col for col in categorical if col not in id_target_col]
print("Initial Numerical:", numerical)
print("Initial Categorical:", categorical)
```

```python
# Detection values
binary_cols = []
ordinal_cols = []
true_numerical = []

for col in df.columns:
    
    if col in id_target_col:
        continue
    
    unique_vals = df[col].nunique()
    
    # Binary
    if unique_vals == 2:
        binary_cols.append(col)
    
    # Categorical (object type OR low unique values)
    elif df[col].dtype == 'object' or unique_vals < 10:
        
        # Possible ordinal (you can adjust manually later)
        ordinal_cols.append(col)
    
    # True numerical
    else:
        true_numerical.append(col)

print(binary_cols)
print(ordinal_cols)
print(true_numerical)
```

 # STEP 3: DATA QUALITY CHECK
       

```python
# check missingvalue
df.isnull().sum()
```

```python
# Handle missing values
for col in numerical:
    null_ratio = df[col].isnull().mean()

    # Drop high missing value
    if null_ratio > 0.3:
        print(f"DROP → {col}: {null_ratio:.2f}")
        df.drop(columns=[col], inplace=True)
        
    # fill moderate missing with median
    elif null_ratio > 0:
        print(f"null_ratio {col} : {null_ratio:.2f}")
        
        median_value = df[col].median()
        print(f"median {col} : {median_value} ")
        
        df[col] = df[col].fillna(df[col].median())
        print(f"Filled missing values in '{col}' with {median_value}")
    print("-----------------------------------")

# check data after hadle missing value
valid_numerical = [col for col in numerical if col in df.columns]
df[valid_numerical].isnull().sum()
```

```python
#Encode binary to a number

# checked binary value
for col in binary_cols:
    print(f"column {col} :", df[col].unique())

    unique_values = df[col].dropna().unique()
    print(f"Value after drop Nan : ", unique_values)

    # Map uniqe number 
    if len(unique_values) == 2:
        if set(unique_values) == {True, False}:
            df[col] = df[col].map({True: 1, False: 0}).astype('Int64')
            print(f"{col} mapping:", {True: 1, False: 0})
            
        else:
            df[col] = df[col].map({unique_values[0]: 0, unique_values[1]: 1})
            print(f"{col} mapping:", {unique_values[0]: 0, unique_values[1]: 1})
            
        # fill 0 for missing value
        df[col] = df[col].fillna(df[col].median())
    
    else:
        print(f"Skipping {col} (not truly binary)")
    print("-----------------------------------")
```
